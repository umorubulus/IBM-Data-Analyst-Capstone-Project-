### Task 1: Visualizing Data Composition with Pie Charts


##### 1.1 Create a Pie Chart of the Top 5 Databases Respondents Want to Work With


In the survey data, the `DatabaseWantToWorkWith` column lists the databases that respondents wish to work with. Let’s visualize the top 5 most-desired databases in a pie chart.



```python
from collections import Counter

# Helper function to plot top 5 pie chart from semicolon-separated columns
def plot_top5_pie(df, column_name, title):
    data = df[column_name].dropna().astype(str).str.split(";")
    counts = Counter(item.strip() for row in data for item in row if item.strip())
    top5 = counts.most_common(5)

    labels = [x[0] for x in top5]
    sizes = [x[1] for x in top5]

    plt.figure(figsize=(7, 7))
    plt.pie(sizes, labels=labels, autopct="%1.1f%%", startangle=140)
    plt.title(title)
    plt.tight_layout()
    plt.show()
```

```python
##Write your code here
plot_top5_pie(
    df,
    "DatabaseWantToWorkWith",
    "Top 5 Databases Respondents Want to Work With"
)

```

The `DevType` column lists the developer types for respondents. We’ll examine the distribution by showing the top 5 developer roles in a pie chart.



```python
##Write your code here
# Task 1.2: Top 5 Developer Roles
plot_top5_pie(
    df,
    "DevType",
    "Top 5 Developer Roles"
)
```
